# **Data Cleaning**

```python
df_anime_copy = df_anime.copy()
df_rating_copy = df_rating.copy()
print("Copy Created Boss✅")
```

```python
df_anime_copy['genre'] = df_anime_copy['genre'].str.split(',')
```

```python
# DataType Correction
df_anime_copy.loc[df_anime_copy['episodes']=='Unknown','episodes'] = np.nan
df_anime_copy['episodes'] = df_anime_copy['episodes'].astype('float')
df_anime_copy.info()
```

```python
# filling NaN values
print('NaN Values Before: ')
print(df_anime_copy.isna().sum())

df_anime_copy['episodes'] = df_anime_copy['episodes'].fillna(round(df_anime_copy['episodes'].median(),0))
df_anime_copy['type'] = df_anime_copy['type'].fillna('Unknown')
df_anime_copy['rating'] = df_anime_copy['rating'].fillna(round(df_anime_copy['rating'].median(),0))
df_anime_copy.explode('genre')
df_anime_copy['genre'] = df_anime_copy['genre'].fillna('Unknown')

print('NaN Values After: ')
df_anime_copy.isna().sum()
```

```python
df_anime_copy.duplicated(subset='anime_id').sum()
```

# **EDA**

**Q1. What are the top 10 highest rated anime of all time?**

```python
plot = df_anime_copy.sort_values(by='rating', ascending=False).head(10)

fig, ax = plt.subplots(figsize=(10,6))  # ✅ correct way

bars = ax.barh(plot['name'], plot['rating'])

# Add rating labels
for bar in bars:
    width = bar.get_width()
    ax.text(width + 0.02,
            bar.get_y() + bar.get_height()/2,
            f"{width:.2f}",
            va='center', fontsize=11, weight='bold')

# Styling
ax.invert_yaxis()
ax.set_title('Top 10 Anime of All Time', fontsize=18, weight='bold', pad=15,loc='left')
ax.set_xlabel('Rating', fontsize=13, weight='bold')

# Clean look
ax.spines[['top','right','left']].set_visible(False)
ax.xaxis.grid(True, linestyle='--', alpha=0.3)

# Better limits & spacing
ax.set_xlim(9, plot['rating'].max() + 0.2)

plt.tight_layout()
plt.show()
```
