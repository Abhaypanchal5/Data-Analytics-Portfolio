# **Data Analysis**

```python
# DataType correction
df_calendar['date'] = pd.to_datetime(df_calendar['date'])
df_sales['order_date'] = pd.to_datetime(df_sales['order_date'])
```

**Q1. What are the total revenue, total cost and total profit across all sales?**

```python
print(f'The Total Revenue: {df_sales['revenue'].sum()}')
print(f'The Total cost: {df_sales['cost'].sum()}')
print(f'The Total Profit: {round(df_sales['profit'].sum(),2)}')
```

**Q2. Which product category generates the highest revenue?**

```python
df_merged = pd.merge(df_product,df_sales,on='product_id',how='inner')
df_merged = df_merged[['category','revenue']]
plot = df_merged.groupby('category')['revenue'].sum().sort_values()

fig, ax = plt.subplots(figsize=(6,6))

ax.pie(
    plot.values,
    labels=plot.index,
    autopct='%1.1f%%',
    startangle=90
)

ax.set_title("Revenue Distribution by Category", fontsize=18, weight='bold', loc='left')

plt.tight_layout()
plt.show()
```

**Q3. How many unique customers have made a purchase?**

```python
df_merged = pd.merge(df_customer,df_sales,on='customer_id',how='inner')
df_merged = df_merged['customer_id'].unique()
print(f'Out of {len(df_customer)}, {len(df_merged)} place an order')
```

**Q4. Which store has the highest total profit — and which city is it in?**

```python
df_merged = pd.merge(df_stores,df_sales,on='store_id',how='inner')
Top = df_merged.groupby('store_id')['profit'].sum().sort_values(ascending=False).head(1)
top_store = df_stores[df_stores['store_id']=='S074']
top_store['Profit'] = Top.values
top_store
```
