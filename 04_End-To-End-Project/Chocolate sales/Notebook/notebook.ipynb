```python
# DataType correction
df_calendar['date'] = pd.to_datetime(df_calendar['date'])
df_sales['order_date'] = pd.to_datetime(df_sales['order_date'])
```

**Q2. Which product category generates the highest revenue?**

```python
df_merged = pd.merge(df_product,df_sales,on='product_id',how='inner')
df_merged = df_merged[['category','revenue']]
plot = df_merged.groupby('category')['revenue'].sum().sort_values()

fig, ax = plt.subplots(figsize=(6,6))

ax.pie(
    plot.values,
    labels=plot.index,
    autopct='%1.1f%%',
    startangle=90
)

ax.set_title("Revenue Distribution by Category", fontsize=18, weight='bold', loc='left')

plt.tight_layout()
plt.show()
```

**Q10. Which product has the best revenue-to-weight ratio? (revenue per gram sold)**

```python
df_merged = pd.merge(df_product, df_sales, on='product_id', how='inner')

df_product_rev = (
    df_merged
    .groupby(['product_id', 'product_name', 'weight_g'])['revenue']
    .sum()
    .reset_index()
)

df_product_rev['revenue_to_Weight'] = round(df_product_rev['revenue']/df_product_rev['weight_g'],3)

# Get top 10
top_products = df_product_rev.sort_values(
    by='revenue_to_Weight', ascending=False
).head(10)

plt.figure(figsize=(10,6))

plt.barh(top_products['product_name'], top_products['revenue_to_Weight'])

plt.title("Top Products by Revenue per Weight", fontsize=20, weight='bold',loc='left')
plt.xlabel("Revenue per Gram",fontsize=15)
plt.ylabel("Product",fontsize=15)

plt.gca().invert_yaxis()  # highest on top
plt.grid(axis='x', linestyle='--', alpha=0.4)
ax=plt.gca()
ax.spines[['top','right']].set_visible(False)
plt.tight_layout()
plt.show()
```

**Q11. "The marketing team wants to run a discount campaign. Find which products have a high discount rate but low profit margin — these are the ones draining revenue."**

```python
# --- Merge data ---
df_merged = pd.merge(df_product, df_sales, on='product_id', how='inner')

# --- Select relevant columns ---
df_merged = df_merged[['product_id','product_name','discount','revenue','profit']]

# --- Aggregate at product level ---
df_product_perf = (
    df_merged
    .groupby(['product_id','product_name'])[['discount','revenue','profit']]
    .mean()
    .reset_index()
)

# --- Calculate profit margin ---
df_product_perf['profit_margin %'] = (
    df_product_perf['profit'] / df_product_perf['revenue'] * 100
)

# --- Define thresholds (quantile-based) ---
high_discount = df_product_perf['discount'].quantile(0.75)
low_margin = df_product_perf['profit_margin %'].quantile(0.25)

# --- Filter problematic products ---
bad_products = df_product_perf[
    (df_product_perf['discount'] >= high_discount) &
    (df_product_perf['profit_margin %'] <= low_margin)
].sort_values(by='profit_margin %')

# --- Plot ---
plt.figure(figsize=(8,6))

# All products
plt.scatter(
    df_product_perf['discount'],
    df_product_perf['profit_margin %'],
    alpha=0.4,
    label='All Products'
)

# Highlight bad products
plt.scatter(
    bad_products['discount'],
    bad_products['profit_margin %'],
    marker='x',
    label='High Discount & Low Margin'
)

# Threshold lines (VERY useful insight)
plt.axvline(high_discount, linestyle='--')
plt.axhline(low_margin, linestyle='--')

plt.xlabel("Discount")
plt.ylabel("Profit Margin %")
plt.title("Discount vs Profit Margin Analysis", fontsize=14, weight='bold')
plt.legend()

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# --- Show top problematic products ---
bad_products.head(10)
```
