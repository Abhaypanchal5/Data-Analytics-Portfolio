# **Data Cleaning**

```python
df_copy = df.copy()
```

```python
# Data have no wrong datatype so we skip this part 

# Outliers Handaling
Outliers = df_copy.select_dtypes(include='number').columns.tolist()

for col in Outliers:
    Q1 = df_copy[col].quantile(0.25)
    Q3 = df_copy[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_copy.loc[
        (df_copy[col] < lower) | (df_copy[col] > upper),
        col
    ] = np.nan
    
print('Outliers Handled Boss✅')
```

```python
NaN_values = df_copy.select_dtypes(include='number').columns.tolist()

for NaN in NaN_values:
    df_copy[NaN] = df_copy[NaN].fillna(round(df_copy[NaN].median(),2))
    
print('NaN values filled with relevent Data',df_copy.isna().sum())
```

```python
df_copy.drop_duplicates(inplace=True)
df_copy.duplicated().sum()
```

# **EDA**

**Q1. What is the overall attrition rate in the company?**

```python
Atteition_rate = df_copy['Attrition'].value_counts()

fig, ax = plt.subplots(figsize=(6,6))

wedges, texts, autotexts = ax.pie(
    Atteition_rate.values,
    labels=Atteition_rate.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width':0.5, 'edgecolor':'black'},
    pctdistance=0.8
)

# Improve text styling
for text in texts:
    text.set_fontsize(12)

for autotext in autotexts:
    autotext.set_fontsize(10)
    autotext.set_weight('bold')

# Center text (nice touch)
ax.text(0, 0, f"Total\n{Atteition_rate.sum()}",
        ha='center', va='center', fontsize=12, weight='bold')

ax.set_title("Attrition rate", fontsize=16, weight='bold')
ax.axis('equal')

plt.tight_layout()
plt.show()
```

**Q2. Which department has the highest attrition?**

```python
df_Dep = (
    df_copy.groupby('Department')['Attrition']
    .value_counts()
    .reset_index(name='count')
)

# Filter ONLY Yes attrition
df_yes = df_Dep[df_Dep['Attrition'] == 'Yes'].sort_values(by='count', ascending=False)

fig, ax = plt.subplots(figsize=(6,6))

wedges, texts, autotexts = ax.pie(
    df_yes['count'],
    labels=df_yes['Department'],
    autopct='%1.1f%%',
    startangle=100,
    radius=1.5,
    wedgeprops={'width':0.6, 'edgecolor':'black'},
    pctdistance=0.8   # better positioning
)

# Improve text styling
for text in texts:
    text.set_fontsize(12)

for autotext in autotexts:
    autotext.set_fontsize(10)
    autotext.set_weight('bold')

# Center text
ax.text(0, 0, f"Total\n{df_yes['count'].sum()}",
        ha='center', va='center', fontsize=12, weight='bold')

ax.set_title("Attrition (Yes) by Department", fontsize=16, weight='bold')
ax.axis('equal')

plt.tight_layout()
plt.show()
```

**Q5. What is the avg monthly income of employees who left vs stayed?**

**Q10. "HR wants to identify the highest-risk employee profile. Find the combination of department, job role, overtime status, and salary bracket that has the highest attrition rate — this is who needs retention intervention first."**

```python
# --- Create Salary Brackets ---
df_temp = df_copy.copy()

df_temp['SalaryBracket'] = pd.cut(
    df_temp['MonthlyIncome'],
    bins=[0, 3000, 6000, 10000, df_temp['MonthlyIncome'].max()],
    labels=['Low', 'Medium', 'High', 'Very High']
)

# --- Compute Attrition Rate ---
risk_profile = (
    df_temp.groupby(['Department','JobRole','OverTime','SalaryBracket'])['Attrition']
    .apply(lambda x: (x == 'Yes').mean())
    .reset_index(name='AttritionRate')
)

# --- Add Count (to avoid small sample bias) ---
counts = df_temp.groupby(['Department','JobRole','OverTime','SalaryBracket']).size().reset_index(name='Count')
risk_profile = risk_profile.merge(counts, on=['Department','JobRole','OverTime','SalaryBracket'])

# Filter meaningful groups
risk_profile = risk_profile[risk_profile['Count'] > 10]

# Get Top 8 risky profiles
top_risk = risk_profile.sort_values(by='AttritionRate', ascending=False).head(8)

# Create readable labels
top_risk['Profile'] = (
    top_risk['Department'] + " | " +
    top_risk['JobRole'] + " | OT:" +
    top_risk['OverTime'] + " | " +
    top_risk['SalaryBracket'].astype(str)
)

# --- Plot ---
fig, ax = plt.subplots(figsize=(12,6))

bars = ax.barh(top_risk['Profile'], top_risk['AttritionRate'])

# Add labels
for bar in bars:
    width = bar.get_width()
    ax.text(width + 0.01,
            bar.get_y() + bar.get_height()/2,
            f"{width:.2f}",
            va='center', fontsize=10, weight='bold')

# Styling
ax.invert_yaxis()
ax.set_xlabel("Attrition Rate", fontsize=12, weight='bold')
ax.set_title("Top High-Risk Employee Profiles", fontsize=16, weight='bold')

ax.spines[['top','right','left']].set_visible(False)
ax.xaxis.grid(True, linestyle='--', alpha=0.4)

ax.set_xlim(0, top_risk['AttritionRate'].max() + 0.1)

plt.tight_layout()
plt.show()
```
