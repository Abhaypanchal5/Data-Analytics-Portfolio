## **Data Cleaning**

```python
df_copy = df.copy()
```

```python
df_copy.info()
```

```python
df.head()
```

### *Filling NaN With Unkown and other factor*

```python
print('NaN values Before:\n',df_copy.isna().sum())

values = {'director':'Unkown',
          'cast':'Unkown',
          'country':'Unkown'}
df_copy.fillna(values,inplace=True)
cols = ['rating', 'date_added', 'duration']

for col in cols:
    df_copy[col] = df_copy[col].fillna(df_copy[col].mode()[0])
    
df_copy.loc[df_copy['rating'].str.contains('min'),'rating'] = 'UR'

print('NaN values After:\n',df_copy.isna().sum())
```

### *DataType Correction*

```python
df_copy['date_added'] = pd.to_datetime(df_copy['date_added'],errors='coerce')
df_copy['listed_in'] = df_copy['listed_in'].str.split(',')
df_copy['cast'] = df_copy['cast'].str.split(',')
df_copy['country'] = df_copy['country'].str.split(',')
```

# **Data Analysis**

```python
df_exploded = df_copy.explode('country')
df_India = df_exploded[df_exploded['country']=='India']
df_India.head()
```

**Q1:Actors with Highest Number of Appearances**

```python
# Globally
df_cast_global = df_copy.explode('cast')
# df_cast_global = df_cast_global[df_cast_global['cast'] != 'Unknown']
df_top_cast_global = df_cast_global['cast'].value_counts().sort_values(ascending=False).head(10)

# Locally (India)
df_cast_india = df_India.explode('cast')
# df_cast_india = df_cast_india[df_cast_india['cast'] != 'Unknown']  
df_top_cast_local = df_cast_india['cast'].value_counts().sort_values(ascending=False).head(10)

fig, axs = plt.subplots(2, 1, figsize=(10, 10), facecolor='#0F0F0F')
fig.suptitle("Actor Appearances", fontsize=26, weight='bold',
             color='white', y=0.98)  # ← was 1.02

datasets = [
    (df_top_cast_global, "Global Top 10", '#E50914'),
    (df_top_cast_local,  "India Top 10",  '#FF9933'),
]

for ax, (data, title, color) in zip(axs, datasets):
    ax.set_facecolor('#1A1A1A')

    bars = ax.bar(data.index, data.values,
                  color=color, edgecolor='none',
                  width=0.6, zorder=3)

    for bar in bars:
        ax.text(bar.get_x() + bar.get_width() / 2,
                bar.get_height() + 0.3,
                str(int(bar.get_height())),
                ha='center', va='bottom',
                fontsize=10, color='white', weight='bold')

    ax.set_title(title, fontsize=16, weight='bold',
                 color='white', loc='left', pad=12)
    ax.set_xlabel("Actors", fontsize=12, color='#AAAAAA', labelpad=8)
    ax.set_ylabel("Appearances", fontsize=12, color='#AAAAAA', labelpad=8)

    ax.set_xticks(range(len(data.index)))
    ax.set_xticklabels(data.index, rotation=40, ha='right',
                       fontsize=9, color='#CCCCCC')

    ax.tick_params(axis='y', colors='#AAAAAA')
    ax.tick_params(axis='x', length=0)

    ax.yaxis.grid(True, linestyle='--', alpha=0.25, color='white', zorder=0)
    ax.set_axisbelow(True)

    for spine in ax.spines.values():
        spine.set_visible(False)

plt.tight_layout(rect=[0, 0, 1, 0.96])  # ← leaves room for suptitle
plt.show()
```

**Q2: How many movies and TV shows we have**

```python
fig, axs = plt.subplots(1, 2, figsize=(12, 6), facecolor='#0F0F0F')

colors = ['#E50914', '#FF9933']

datasets = [
    (df_copy['type'].value_counts(),  "Global"),
    (df_India['type'].value_counts(), "India"),
]

for ax, (data, title) in zip(axs, datasets):
    ax.set_facecolor('#1A1A1A')

    wedges, texts, autotexts = ax.pie(
        data.values,
        labels=data.index,
        autopct=lambda pct: f"{int(round(pct/100 * sum(data.values)))}",
        colors=colors,
        startangle=90,
        wedgeprops=dict(edgecolor='#0F0F0F', linewidth=3)
    )

    for text in texts:
        text.set_color('white')
        text.set_fontsize(13)
        text.set_fontweight('bold')

    for autotext in autotexts:
        autotext.set_color('white')
        autotext.set_fontsize(12)
        autotext.set_fontweight('bold')

    ax.set_title(title, fontsize=18, weight='bold', color='white', pad=20)

fig.suptitle("Movies vs TV Shows", fontsize=24, weight='bold',
             color='white', y=1.02)

plt.tight_layout()
plt.show()
```

**Q3: Movies with the minimum lenght of 2h**

```python
df_movie = df_copy[df_copy['type'] == 'Movie'].copy()
df_movie['duration_min'] = df_movie['duration'].str.replace(' min', '')
df_movie = df_movie[~df_movie['duration_min'].str.contains('Season', na=False)]
df_movie['duration_min'] = df_movie['duration_min'].astype('int')

df_working_movies = df_movie[df_movie['duration_min'] >= 120]
df_working_movies = df_working_movies.sort_values(ascending=False, by='duration_min').head(10)

# ── Plot ───────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 6), facecolor='#0F0F0F')
ax.set_facecolor('#1A1A1A')

bars = ax.bar(df_working_movies['title'], df_working_movies['duration_min'],
              width=0.6, color='#E50914', edgecolor='none', zorder=3)

# value labels
for bar in bars:
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 2,
            f"{int(bar.get_height())} min",
            ha='center', va='bottom',
            fontsize=9, color='white', weight='bold')

# 2h reference line
ax.axhline(y=120, color='#FF9933', linewidth=1.5,
           linestyle='--', zorder=4, label='2h mark (120 min)')
ax.legend(facecolor='#2A2A2A', edgecolor='none',
          labelcolor='white', fontsize=10)

ax.set_title("Top 10 Longest Movies  (min. 2h)", fontsize=18,
             weight='bold', color='white', loc='left', pad=15)
ax.set_xlabel("Movie", fontsize=12, color='#AAAAAA', labelpad=8)
ax.set_ylabel("Duration (min)", fontsize=12, color='#AAAAAA', labelpad=8)

ax.set_xticks(range(len(df_working_movies)))
ax.set_xticklabels(df_working_movies['title'].values,
                   rotation=35, ha='right', fontsize=9, color='#CCCCCC')

ax.tick_params(axis='y', colors='#AAAAAA')
ax.tick_params(axis='x', length=0)
ax.yaxis.grid(True, linestyle='--', alpha=0.25, color='white', zorder=0)
ax.set_axisbelow(True)

for spine in ax.spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()
```

**Q2. Which country has produced the most content on Netflix?**

```python
df_country = df_copy.copy()
df_country = df_country.explode('country')
df_country['country'] = df_country['country'].str.strip()

top10 = df_country['country'].value_counts().head(10)

fig, ax = plt.subplots(figsize=(10, 6), facecolor='#0F0F0F')
ax.set_facecolor('#1A1A1A')

bars = ax.barh(top10.index, top10.values, color='#E50914', edgecolor='#0F0F0F')
ax.invert_yaxis()

# Value labels on bars
for bar in bars:
    ax.text(bar.get_width() + 10, bar.get_y() + bar.get_height()/2,
            str(int(bar.get_width())),
            va='center', color='white', fontsize=11, fontweight='bold')

ax.set_title("Top 10 Countries by Content Volume", fontsize=18,
             weight='bold', color='white', loc='left', pad=15)
ax.set_xlabel('Number of Titles', fontsize=15, color='white')
ax.tick_params(colors='white')
ax.xaxis.grid(True, linestyle='--', alpha=0.3, color='white')
ax.spines[['top','right']].set_visible(False)

plt.tight_layout()
plt.show()
```

**Q3. What are the top 10 most common ratings (PG, TV-MA etc)?**

```python
top10_ratings = df_copy['rating'].value_counts().head(10)
avg_count = top10_ratings.values.mean()

fig, ax = plt.subplots(figsize=(10, 6), facecolor='#0F0F0F')
ax.set_facecolor('#1A1A1A')

bars = ax.barh(top10_ratings.index, top10_ratings.values,
               color='#E50914', edgecolor='#0F0F0F')
ax.invert_yaxis()

# Value labels
for bar in bars:
    ax.text(bar.get_width() + 10, bar.get_y() + bar.get_height()/2,
            str(int(bar.get_width())),
            va='center', color='white', fontsize=11, fontweight='bold')

# Average line across top 10
ax.axvline(x=avg_count, color='#FF9933', linewidth=2,
           linestyle='--', zorder=10, label=f'Avg: {int(avg_count)}')

ax.set_title("Top 10 Most Common Content Ratings", fontsize=18,
             weight='bold', color='white', loc='left', pad=15)
ax.set_xlabel('Number of Titles', fontsize=13, color='white')
ax.tick_params(colors='white')
ax.xaxis.grid(True, linestyle='--', alpha=0.3, color='white')
ax.spines[['top','right','left','bottom']].set_visible(False)
ax.legend(fontsize=12, facecolor='#1A1A1A', labelcolor='white')

plt.tight_layout()
plt.show()
```
