# **Data Cleaning**

## *Create Copy*

```python
df_copy = df.copy()
print('Copy Created Boss✅')
```

## *Cleaning*

```python
# Name Column
replacments = {'@':'A','3':'E','0':'O'} 
for char, replacment in replacments.items():
    df_copy['driver_name'] = df_copy['driver_name'].replace(char,replacment,regex=True)
df_copy['driver_name'] = df_copy['driver_name'].str.strip().str.title()



# Age Column
replacments = {'yrs':'','unknown':np.nan} 
for char, replacment in replacments.items():
    df_copy['age'] = df_copy['age'].replace(char,replacment,regex=True)
df_copy['age'] = df_copy['age'].str.strip().astype('float')
df_copy.loc[df_copy['age']<18,'age'] = np.nan
df_copy.loc[df_copy['age']>50,'age'] = np.nan



# City Column
df_copy['city'] = df_copy['city'].str.strip().str.title()
replacments = {'Chenna':'Chennai','Bangalroe':'Bangalore','Dlhi':'Delhi','Mumabi':'Mumbai','Hydrabad':'Hyderabad'}
for char,replacment in replacments.items():
    df_copy['city'] = df_copy['city'].replace(char,replacment)
    
    
    
# Phone No. Column
df_copy['phone'] = df_copy['phone'].str.replace(r'\+91|-|\s', '', regex=True)
df_copy.loc[(df_copy['phone'].str.len() != 10),'phone'] = 'Unknown'




# Email Column
df_copy['email'] = df_copy['email'].replace('@@','@',regex=True)
df_copy['email'] = df_copy['email'].fillna('Unknown')




# Platform,Vehical Type & Condition, shift, Experience Column
columns = ['vehicle_type',
           'platform',
           'vehicle_condition',
           'experience_level',
           'preferred_shift',
           'reason_for_joining',
           'document_verification',
           'background_check_status',
           'interview_result',
           'rejection_reason']
for col in columns:
    df_copy[col] = df_copy[col].str.strip().str.title()
    replacments = {'None':'Unknown','-':'Unknown','Na':'Unknown','Avg':'Average'}
    df_copy[col] = df_copy[col].replace(replacments)
    df_copy[col] = df_copy[col].fillna('Unknown')
    
    
    

# Language Column
df_copy['languages_known'] = df_copy['languages_known'].str.split(',')



# Rating Column
df_copy['overall_driver_rating'] = df_copy['overall_driver_rating'].replace('unrated',np.nan).astype('float')
df_copy.loc[df_copy['overall_driver_rating']<0,'overall_driver_rating'] = np.nan
df_copy.loc[df_copy['overall_driver_rating']>10,'overall_driver_rating'] = np.nan
df_copy['overall_driver_rating'] = df_copy['overall_driver_rating'].fillna(-1) # -1 = Unrated



# Trip Column
df_copy.loc[df_copy['total_trips_completed']<0,'total_trips_completed'] = np.nan
df_copy['total_trips_completed'] = df_copy['total_trips_completed'].fillna(round(df_copy['total_trips_completed'].median(),0))



# Cancellation Rate
df_copy.loc[df_copy['cancellation_rate_%']<0,'cancellation_rate_%'] = np.nan
df_copy.loc[df_copy['cancellation_rate_%']>100,'cancellation_rate_%'] = np.nan
df_copy['cancellation_rate_%'] = df_copy['cancellation_rate_%'].fillna(round(df_copy['cancellation_rate_%'].mean(),1))




# Salary Column
df_copy['avg_daily_earnings_INR'] = df_copy['avg_daily_earnings_INR'].replace(['₹','Rs.','/-'],'',regex=True).astype('float')
df_copy.loc[df_copy['avg_daily_earnings_INR']<0,'avg_daily_earnings_INR'] = np.nan
df_copy.loc[df_copy['avg_daily_earnings_INR']>10000,'avg_daily_earnings_INR'] = np.nan
df_copy['avg_daily_earnings_INR'] = df_copy['avg_daily_earnings_INR'].fillna(round(df_copy['avg_daily_earnings_INR'].mean(),1))




# Complaint, Compliments column
columns = ['total_compliments','total_complaints']
for col in columns:
    df_copy.loc[df_copy[col]<0,col] = np.nan
    df_copy.loc[df_copy[col]>50,col] = np.nan
    df_copy[col] = df_copy[col].fillna(round(df_copy[col].median(),0))
    
    
    
    
# Availablity Column
column = ['peak_hour_availability','is_currently_active']
for items in column:
    df_copy[items] = df_copy[items].str.strip().str.title().replace(['False','Na','0','N'],'No')
    df_copy[items] = df_copy[items].str.strip().str.title().replace(['1','True','Y'],'Yes')
    df_copy[items] = df_copy[items].str.strip().str.title().replace(['-','None'],'Unknown')
    df_copy[items] = df_copy[items].fillna('Unknown')
    
    
    

# Date Column
import warnings

with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    columns = ['interview_date','onboarding_date']
    for col in columns:
        df_copy[col] = pd.to_datetime(
            df_copy[col],
            errors='coerce',
            dayfirst=True,
            utc=True
        )
        df_copy[col] = df_copy[col].fillna(df_copy[col].median())





# Communication Score
columns = ['driving_test_score_10',
           'communication_score_10',
           'route_knowledge_score_10',
           'customer_handling_score_10',
           'overall_interview_score_10']
for col in columns:
    df_copy.loc[df_copy[col]<0,col] = np.nan
    df_copy.loc[df_copy[col]>10,col] = np.nan
    df_copy[col] = df_copy[col].fillna(round(df_copy[col].mean(),1))
```

## *NaN Values*

```python
print('NaN Values Before: \n')
print(df_copy.isna().sum())
df_copy['driver_id'] = df_copy['driver_id'].fillna('Unknown')
df_copy['age'] = df_copy['age'].fillna(round(df_copy['age'].median(),0))
df_copy['languages_known'] = df_copy['languages_known'].fillna('Unknown')
df_copy['trip_completion_rate_%'] = df_copy['trip_completion_rate_%'].fillna(round(df_copy['trip_completion_rate_%'].median(),2))
print('NaN Values After: \n')
print(df_copy.isna().sum())
```

## *Drop Duplicates*

```python
columns = df_copy.columns.tolist()
columns.remove('languages_known')
print(f'Duplicates in the data before is {df_copy[columns].duplicated().sum()}')
df_copy.drop_duplicates(inplace=True,subset=columns)
print(f'Duplicates in the data After is {df_copy[columns].duplicated().sum()}')
```
