## Import the neccesory libraries

```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine
```

## Load the data 

```python
df = pd.read_csv("C:\\Users\\abhay\\OneDrive\\Desktop\\Data-Analytics-Portfolio\\05_End-To-End-Project\\Ecommerce_sales_analysis_avd\\Data\\ecommerce_sales.csv")
df["order_date"] = pd.to_datetime(df["order_date"])
df.head()
```

## Data understanding

- NaN values

```python
print("The NaN values in the DataFrame is as follow")
print(df.isna().sum())
```

- Duplicates

```python
print("The number of dulates in the Data is",df.duplicated().sum())
```

- Data frame shape and info

```python
df.info()
df.shape
```

- Filling the age NaN values

```python
df["age"] = df["age"].fillna(df["age"].mean())
```

- Filling NaN values of customer name

```python
df["customer_name"] = df["customer_name"].fillna("Unkown")
```

- Filling NaN values of rating

```python
df["rating"] = df["rating"].fillna(df["rating"].mean())
```

- Drop the duplicate values

```python
df.drop_duplicates(inplace=True)
```

## Create a cleaned data file

```python
df.to_csv("cleaned_ecommerce_data.csv",index=False)
```
