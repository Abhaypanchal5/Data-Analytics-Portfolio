## ANALYSIS

Analysis based on Attrition

```python
# Total employee 
print(df["EmployeeID"].count())

# group the employee based on stay or not 
stay_number = df.groupby("Attrition")["Attrition"].count()

# no of employee stay 
print(stay_number["No"])

# no of employee leave 
print(stay_number["Yes"])

# % of Employee leave the company
print(stay_number["Yes"]/df["EmployeeID"].count()*100)

```

Attrition rate based on department

```python
# group the department based on attrition 
dep_rate = df.groupby("Department")["Attrition"].value_counts()

# Top department by attrition number 
rate = dep_rate[:,"Yes"].sort_values(ascending=False)

# rate of leaving by the department 
dept_total = df.groupby("Department")["EmployeeID"].count()
rate/dept_total * 100 
```

```python
# What is the average monthly income of employees who left vs stayed? 
print(df.groupby("Attrition")["MonthlyIncome"].mean() )

# Do employees who work overtime leave more often?
overtime_rate = df.groupby("OverTime")["Attrition"].value_counts()

overtime_rate[:,"Yes"].sort_values(ascending=False)
```
