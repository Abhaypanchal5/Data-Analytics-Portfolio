## ANALYSIS

Analysis based on Attrition

```python
# Total employee 
df["EmployeeID"].count()

# group the employee based on stay or not 
stay_number = df.groupby("Attrition")["Attrition"].count()

# no of employee stay 
stay_number["Yes"]

# no of employee leave 
stay_number["No"]

# % of Employee leave the complany 
stay_number["No"]/df["EmployeeID"].count()*100

```

Attriction rate based on department

```python
# group the department based on attrition 
dep_rate = df.groupby("Department")["Attrition"].value_counts()

# Top department by attrition number 
rate = dep_rate[:,"No"].sort_values(ascending=False)

# rate of leaving by the department 
rate/df["EmployeeID"].count()*100
```
